```python
import pandas as pd

df = pd.read_csv("../Data/2016_Financial_Data.csv")

print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print(f"Antal företag: {df.shape[0]}")
print(f"Antal variabler: {df.shape[1]}")

print("\nFörsta 5 raderna:")
display(df.head())

print("\nKolumner:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python
print("=" * 60)
print("DATATYPER")
print("=" * 60)

display(df.dtypes.to_frame(name="Datatyp"))

print("\nAntal kolumner per datatyp:")
print(df.dtypes.value_counts())
```

```python
print("=" * 60)
print("SAKNADE VÄRDEN")
print("=" * 60)

missing = pd.DataFrame({
    "Antal saknade": df.isna().sum(),
    "Andel saknade (%)": df.isna().mean() * 100
})

missing = missing.sort_values("Andel saknade (%)", ascending=False)

display(missing.head(30))
```
