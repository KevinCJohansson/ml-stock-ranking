## 1. Dataöversikt

Jag undersöker 2015 års finansiella data för att få en överblick över datasetet.

Resultat:

Datasetet innehåller 4 120 rader och 225 kolumner.

```python
import pandas as pd

df = pd.read_csv("../Data/2015_Financial_Data.csv")

print("Antal rader och kolumner:", df.shape)
display(df.head())
```

## 2. Datakvalitet – saknade värden

Jag undersöker vilka kolumner som har saknade värden och visar de 15 med flest.

```python
# Kontrollera saknade värden per kolumn
missing_values = df.isnull().sum()

# Visa de 15 kolumner som har flest saknade värden
display(missing_values.sort_values(ascending=False).head(15))
```

Resultat:

Flera kolumner har saknade värden.

- operatingCycle: 4 118 saknade värden
- cashConversionCycle: 4 118 saknade värden

Dessa två kolumner saknar nästan all data.
Inga värden har ändrats eller tagits bort.
