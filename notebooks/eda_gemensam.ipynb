# Gemensam EDA – Finansiella nyckeltal 2014–2018

## Syfte

Denna notebook sammanställer gruppens utforskande dataanalys (EDA) av finansiella data för amerikanska aktier under åren 2014–2018.

Syftet är att undersöka datamängdens struktur, saknade värden, fördelningar och extremvärden samt identifiera eventuella skillnader mellan åren.

EDA:n utgår från gruppmedlemmarnas tidigare analyser och fungerar som underlag för projektet "Kan finansiella nyckeltal slå marknaden?"

## 1. Sammanfattning av EDA 2014–2018

I detta avsnitt sammanfattar vi de viktigaste resultaten från gruppens fem EDA-analyser. Vi undersöker datakvalitet, saknade värden, extremvärden och finansiella nyckeltal samt jämför eventuella skillnader mellan åren.

### EDA 2014

Analysen av 2014 års data visade flera viktiga kvalitetsproblem:

- 186 bolag saknade mer än 50 % av sina värden.
- Det förekom extrema och orimliga avkastningsvärden som kan bero på datafel.
- P/E-talet verkar vara satt till 0 för bolag med negativ vinst, vilket behöver beaktas.
- Avkastningen varierade tydligt mellan olika sektorer.
- En målvariabel, "above_median", skapades för att visa vilka aktier som överträffade medianavkastningen under 2015.

Slutsats:

Datan behöver kvalitetskontrolleras och extremvärden hanteras innan modellering. Kolumnen "Class" ska inte användas som en förklarande variabel.

### EDA 2015

Analysen av 2015 års data visade följande:

- Datasetet innehåller 4 120 företag och 225 kolumner.
- De flesta kolumnerna innehåller numeriska värden.
- Kolumnerna operatingCycle och cashConversionCycle saknar vardera 4 118 värden av totalt 4 120.
- Kolumnen shortTermCoverageRatios saknar 1 925 värden.
- Datasetet innehåller information om företagens sektorer.
- Financial Services är den största sektorn med 769 företag, följt av Healthcare med 628 och Technology med 606 företag.

Slutsats:

Analysen visar att flera finansiella nyckeltal har stora mängder saknade värden. Framför allt saknar två kolumner nästan all information. Detta behöver beaktas vid datarensningen och jämföras med övriga år innan gruppen bestämmer vilka variabler som ska användas i modellen.

### EDA 2016

Analysen av 2016 års data visade följande:

- Datasetet innehåller 4 797 företag och 225 kolumner.
- Av kolumnerna är 222 numeriska med decimaltal, två innehåller text och en innehåller heltal.
- Kolumnerna operatingCycle och cashConversionCycle saknar vardera 4 792 värden, vilket motsvarar cirka 99,9 %.
- Kolumnen shortTermCoverageRatios saknar 2 417 värden, cirka 50,4 %.
- Även 10Y Shareholders Equity Growth (per Share) har många saknade värden, cirka 47,7 %.
- Financial Services är den största sektorn med 1 220 företag, följt av Healthcare med 686 och Technology med 643 företag.

Slutsats:

Precis som under 2015 finns stora mängder saknade värden i vissa finansiella nyckeltal. Framför allt saknar operatingCycle och cashConversionCycle nästan all information. Det tyder på att dessa variabler kan vara mindre lämpliga för modellering utan särskild hantering.

Analysen visar också att företagen är ojämnt fördelade mellan sektorerna, vilket är relevant att ta hänsyn till vid tolkningen av resultaten.

### EDA 2017

Analysen av 2017 års data visade följande:

- Datasetet innehåller 4 960 företag och 225 kolumner.
- Inga dubbletter identifierades.
- Kolumnerna operatingCycle, cashConversionCycle och shortTermCoverageRatios har över 50 % saknade värden.
- Den första kolumnen, Unnamed: 0, innehåller företagens börskoder.
- Datasetet innehåller 11 sektorer och inga saknade värden i sektorkolumnen.
- Kolumnen 2018 PRICE VAR [%] saknar inga värden.
- En målvariabel, Target, skapades för att klassificera aktier som över eller under medianavkastningen. Fördelningen är ungefär 50/50.

Slutsats:

Även under 2017 finns stora mängder saknade värden i vissa finansiella nyckeltal. Samtidigt saknas inga värden i avkastningskolumnen som används för att skapa målvariabeln.

Det är viktigt att hantera saknade värden innan modellering och att inte använda framtida avkastning som en förklarande variabel.

### EDA 2018

Analysen av 2018 års data visade följande:

- Datasetet innehåller 4 392 företag och 225 kolumner.
- Av kolumnerna är 222 numeriska med decimaltal, två innehåller text och en innehåller heltal.
- Kolumnerna operatingCycle och cashConversionCycle saknar vardera 4 386 värden, cirka 99,9 %.
- Kolumnen shortTermCoverageRatios saknar 1 926 värden, cirka 43,9 %.
- Även 10Y Shareholders Equity Growth (per Share) har många saknade värden, cirka 38,6 %.
- Financial Services är den största sektorn med 824 företag, följt av Healthcare med 691 och Technology med 636 företag.
- Datasetet innehåller kolumnen 2019 PRICE VAR [%], som beskriver den efterföljande avkastningen.
- Den statistiska översikten visar stora skillnader mellan minimi- och maximivärden för vissa finansiella nyckeltal.

Slutsats:

Precis som under tidigare år finns stora mängder saknade värden i vissa finansiella nyckeltal. De stora skillnaderna mellan minimi- och maximivärden tyder också på att extrema värden behöver undersökas.

Eftersom 2018 är projektets slutliga testår ska resultaten från detta år inte användas för att välja variabler eller anpassa modellen. Däremot kan datans struktur och kvalitet beskrivas som en del av EDA:n.

## 2. Jämförelse mellan åren 2014–2018

### Antal företag

Antalet företag i datasetet varierar mellan åren.

2014: 3 808 företag
2015: 4 120 företag
2016: 4 797 företag
2017: 4 960 företag
2018: 4 392 företag

Antalet företag ökade från 2014 till 2017, men minskade under 2018. Det innebär att datasetets storlek förändras över tid, vilket är viktigt att känna till när vi jämför resultaten från olika år.

Alla fem åren innehåller 225 kolumner.

### Saknade värden

När vi jämför åren ser vi att flera finansiella nyckeltal har många saknade värden.

Ett problem som återkommer varje år är kolumnerna operatingCycle och cashConversionCycle.

- 2014: Båda kolumnerna saknar 3 807 av 3 808 värden, cirka 99,97 %.
- 2015: Båda kolumnerna saknar 4 118 av 4 120 värden, cirka 99,95 %.
- 2016: Båda kolumnerna saknar 4 792 av 4 797 värden, cirka 99,90 %.
- 2017: Båda kolumnerna saknar 4 955 av 4 960 värden, cirka 99,90 %.
- 2018: Båda kolumnerna saknar 4 386 av 4 392 värden, cirka 99,86 %.

Även shortTermCoverageRatios har många saknade värden under flera år.

Slutsats:

Vi ser att problemet med saknade värden finns under alla fem åren och inte bara under ett enskilt år.

Vi behöver hantera de saknade värdena på samma sätt för alla år, så att vi kan jämföra modellerna på ett rättvist sätt.

### Sektorfördelning

Vi har jämfört de tre största sektorerna mellan 2014 och 2018.

2014:
- Financial Services: 660 företag
- Healthcare: 582 företag
- Technology: 576 företag

2015:
- Financial Services: 769 företag
- Healthcare: 628 företag
- Technology: 606 företag

2016:
- Financial Services: 1 220 företag
- Healthcare: 686 företag
- Technology: 643 företag

2017:
- Financial Services: 1 247 företag
- Healthcare: 718 företag
- Technology: 665 företag

2018:
- Financial Services: 824 företag
- Healthcare: 691 företag
- Technology: 636 företag

Slutsats:

Financial Services är den största sektorn under alla fem åren. Antalet företag ökade fram till 2017 men minskade under 2018.

Healthcare och Technology har också många företag, men förändringarna mellan åren är mindre.

Eftersom vissa sektorer har fler företag än andra kan de få större påverkan på modellens resultat.

## 3. Gemensam slutsats

När vi jämför åren 2014–2018 ser vi flera likheter.

- Antalet företag varierar mellan åren.
- Vissa finansiella nyckeltal har väldigt många saknade värden.
- Financial Services är den största sektorn under alla fem åren.
- Det finns stora skillnader mellan vissa finansiella värden, vilket kan tyda på extremvärden.

Innan vi går vidare med modellerna behöver vi se över hur vi ska hantera saknade värden och extremvärden. Vi har också sett att vissa sektorer har fler företag än andra, vilket kan påverka resultaten.

Jämförelsen mellan åren har gett oss en bättre bild av datan och vad vi behöver tänka på när vi fortsätter med modelleringen.
