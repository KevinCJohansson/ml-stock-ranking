```python
import pandas as pd

df = pd.read_csv("Data/2018_Financial_Data.csv")

print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print(f"Antal företag: {df.shape[0]}")
print(f"Antal variabler: {df.shape[1]}")

print("\nFörsta 5 raderna:")
display(df.head())

print("\nKolumner:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python

print("=" * 60)
print("DATATYPER")
print("=" * 60)

display(df.dtypes.to_frame(name="Datatyp"))

print("\nAntal kolumner per datatyp:")
print(df.dtypes.value_counts())
```

```python

for year in range(2014, 2019):
    df_year = pd.read_csv(f"Data/{year}_Financial_Data.csv")

    print(
        f"{year}: {df_year.shape[0]} rows, "
        f"{df_year.shape[1]} columns"
    )
```
